```python
from langchain.agents import create_agent
from langchain_groq import ChatGroq

model = ChatGroq(model = "openai/gpt-oss-20b", temperature = 0)
```

```python
agent = create_agent(model = model, system_prompt="You're are chatty genz assistant")
```

```python
response = agent.invoke({"messages" : [{"role" : "user", "content" : "What is the capital of france?"}]})
```

```python
print(response["messages"][-1].text)
```

```python
from pydantic import BaseModel, Field
from typing import List
from typing import TypedDict
from typing import Annotated


class SportsPerson(BaseModel):
    name:str = Field(..., description="name of the person")
    sports:str = Field(..., description="sports associated with the person")
    acheivements:str = Field(..., description="What has the sportsperson acheived")
    trivia:List[str] = Field(..., description="3-5 trivias about the person")

class SportsPerson1(TypedDict):
    name:Annotated[str, Field(..., description="name of the person")]
    sports:Annotated[str, Field(..., description="sports associated with the person")]
    acheivements:Annotated[str, Field(..., description="What has the sportsperson acheived")]
    trivia:Annotated[str, Field(..., description="3-5 trivias about the person")]
```

```python
agent = create_agent(model = model, response_format=SportsPerson1)
response = agent.invoke({"messages" : {'role' : "user", "content" : "Give me details about MS Dhoni"}})
```

```python
sportsperson = response["structured_response"]
```

```python
from langchain.agents import create_agent
from langchain.agents.middleware import SummarizationMiddleware
from langgraph.checkpoint.memory import InMemorySaver

agent = create_agent(model = model, checkpointer = InMemorySaver(), middleware=[SummarizationMiddleware(
    model = model,
    trigger = ("tokens" , 500),
    keep = ("messages", 4)
)])
```

```python
def query(content):
    return {"messages" : [{"role" : "user", "content" : content}]}
```

```python
config = {"configurable" : {"thread_id" : "1234"}}
```

```python
query1 = query("Write a 7000 word essay on Yourself")
```

```python
response = agent.invoke(query1, config = config)
```

```python
response1 = agent.invoke(
    query("Explain global warming in detail."),
    config=config
)

response2 = agent.invoke(
    query("Explain its effects on India."),
    config=config
)

response3 = agent.invoke(
    query("Suggest ways to reduce these effects."),
    config=config
)

print(response3["messages"])
```
